````python
import socket
import datetime
import traceback
import pem
from OpenSSL import SSL, crypto
from cryptography.x509.oid import ExtensionOID, NameOID
import cryptography.x509


def get_server_certificates(hostname, port=443):
    """
    Downloads the leaf certificate and all intermediate certificates from the server.
    Uses pyOpenSSL so it works reliably on all Python versions.
    """
    ctx = SSL.Context(SSL.TLS_CLIENT_METHOD)
    sock = socket.create_connection((hostname, port), timeout=10)
    sock.setblocking(True)

    conn = SSL.Connection(ctx, sock)
    conn.set_connect_state()
    conn.set_tlsext_host_name(hostname.encode("utf-8"))
    conn.do_handshake()

    # Get the certificate chain sent by the server
    chain = conn.get_peer_cert_chain()
    leaf_pem = crypto.dump_certificate(crypto.FILETYPE_PEM, chain[0]).decode("utf-8")
    intermediate_pems = [
        crypto.dump_certificate(crypto.FILETYPE_PEM, cert).decode("utf-8")
        for cert in chain[1:]
    ]

    conn.close()
    sock.close()
    return leaf_pem, intermediate_pems


def verify_chain_of_trust(cert_pem, intermediate_pems, root_pems):
    """
    Validates a leaf certificate against root CA certificates,
    supplying intermediate certificates to complete the chain of trust.
    """
    certificate = crypto.load_certificate(crypto.FILETYPE_PEM, cert_pem)
    store = crypto.X509Store()

    # Add trusted root certificates
    for root_pem in root_pems:
        try:
            trusted_cert = crypto.load_certificate(crypto.FILETYPE_PEM, str(root_pem))
            store.add_cert(trusted_cert)
        except Exception:
            continue

    # Prepare intermediate certificates sent by the server
    intermediates = []
    for int_pem in intermediate_pems:
        try:
            int_cert = crypto.load_certificate(crypto.FILETYPE_PEM, str(int_pem))
            intermediates.append(int_cert)
        except Exception:
            continue

    # Supply intermediate certificates via the chain parameter
    store_ctx = crypto.X509StoreContext(store, certificate, chain=intermediates)
    try:
        store_ctx.verify_certificate()
        return True, "Certificate chain verified successfully"
    except crypto.X509StoreContextError as e:
        return False, f"Verification failed: {e}"


def verify_validity_period(cert_pem):
    """
    Checks if the certificate is currently valid (not expired, already active).
    """
    cert = crypto.load_certificate(crypto.FILETYPE_PEM, cert_pem)
    c_cert = cert.to_cryptography()
    now = datetime.datetime.now(datetime.timezone.utc)

    not_before = c_cert.not_valid_before_utc
    not_after = c_cert.not_valid_after_utc

    if now < not_before:
        return False, f"Certificate is not yet valid (valid from {not_before})"
    if now > not_after:
        return False, f"Certificate has expired (expired on {not_after})"

    return True, f"Valid from {not_before} to {not_after}"


def verify_hostname(cert_pem, hostname):
    """
    Validates that the certificate matches the target hostname,
    checking Subject Alternative Names (SAN) and Common Name (CN),
    supporting wildcard certificates (e.g. *.chula.ac.th).
    """
    cert = crypto.load_certificate(crypto.FILETYPE_PEM, cert_pem)
    c_cert = cert.to_cryptography()

    names = []
    try:
        san = c_cert.extensions.get_extension_for_oid(ExtensionOID.SUBJECT_ALTERNATIVE_NAME)
        names.extend(san.value.get_values_for_type(cryptography.x509.DNSName))
    except cryptography.x509.ExtensionNotFound:
        pass

    # Fallback to CN if no SAN is present
    if not names:
        for attr in c_cert.subject.get_attributes_for_oid(NameOID.COMMON_NAME):
            names.append(attr.value)

    target = hostname.lower()
    for pattern in names:
        pat = pattern.lower()
        if pat.startswith("*."):
            suffix = pat[2:]
            target_parts = target.split(".")
            suffix_parts = suffix.split(".")
            # Single-level wildcard match (RFC 6125)
            if len(target_parts) == len(suffix_parts) + 1 and target.endswith("." + suffix):
                return True, f"Matches wildcard '{pattern}'"
        elif pat == target:
            return True, f"Matches '{pattern}'"

    return False, f"Hostname mismatch: '{hostname}' not in {names}"


def validate_server_certificate(hostname, port=443, ca_file="./ca-certificates.crt"):
    """
    Performs full certificate validation:
    1. Fetches leaf and intermediate certificates
    2. Verifies cryptographic chain of trust to root CAs
    3. Checks validity period
    4. Validates hostname
    """
    pems = pem.parse_file(ca_file)
    root_pems = [str(p) for p in pems]

    leaf_pem, intermediate_pems = get_server_certificates(hostname, port)

    # 1. Chain of trust
    chain_ok, chain_msg = verify_chain_of_trust(leaf_pem, intermediate_pems, root_pems)
    if not chain_ok:
        return False, f"Chain of Trust: {chain_msg}"

    # 2. Validity period
    val_ok, val_msg = verify_validity_period(leaf_pem)
    if not val_ok:
        return False, f"Validity: {val_msg}"

    # 3. Hostname verification
    host_ok, host_msg = verify_hostname(leaf_pem, hostname)
    if not host_ok:
        return False, f"Hostname: {host_msg}"

    return True, f"{chain_msg} | {val_msg} | {host_msg}"


def run_verification(targets, ca_file="./ca-certificates.crt"):
    try:
        pems = pem.parse_file(ca_file)
        print(f"Loaded {len(pems)} root certificates from {ca_file}")
    except Exception as e:
        print(f"[FATAL] Could not read CA file '{ca_file}': {e}")
        return

    for target in targets:
        print(f"\n--- Verifying: {target} ---")
        try:
            success, msg = validate_server_certificate(target, ca_file=ca_file)
            if success:
                print(f"[OK] {target}: {msg}")
            else:
                print(f"[FAIL] {target}: {msg}")
        except Exception as e:
            print(f"[ERROR] Could not process {target}:")
            traceback.print_exc()


if __name__ == "__main__":
    targets = [
        "twitter.com",
        "google.com",
        "www.chula.ac.th",
        "classdeedee.cloud.cp.eng.chula.ac.th"
    ]
    run_verification(targets)
````
